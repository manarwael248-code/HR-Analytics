### **Data Cleaning**

```python
df.isnull().sum()
```

```python
df.dropna(inplace= True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace= True)
df.duplicated().sum()
```

```python
cols_to_drop = ['EmployeeCount', 'Over18', 'StandardHours', 'EmployeeNumber']
df.drop(columns= cols_to_drop, inplace= True)
df.shape
```

```python
cont_cols = ['Age' , 'DailyRate' , 'DistanceFromHome' , 'HourlyRate' , 'MonthlyIncome' , 'MonthlyRate' , 'NumCompaniesWorked' ,
             'PercentSalaryHike' , 'TotalWorkingYears' , 'TrainingTimesLastYear' , 'YearsAtCompany' , 'YearsInCurrentRole' ,
             'YearsSinceLastPromotion' , 'YearsWithCurrManager']
```

```python
plt.figure(figsize= (18, 10))
for i, col in enumerate(cont_cols):
    plt.subplot(3, 5, i + 1)
    sns.boxplot(y= df[col])
plt.tight_layout()
plt.show()
```

```python
for col in cont_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    outliers = df[(df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)].shape[0]
    if outliers > 0:
        print(col , outliers)
```

### **Data Transformation**

```python
df['Attrition_num'] = df['Attrition'].map({'Yes': 1 , 'No': 0})
```

```python
df['AgeGroup'] = pd.cut(df['Age'] , bins= [17 , 25 , 35 , 45 , 55 , 60] , labels= ['18-25' , '26-35' , '36-45' , '46-55' , '56-60'])
```

```python
df['IncomeGroup'] = pd.qcut(df['MonthlyIncome'] , q= 4 , labels= ['Low' , 'Medium' , 'High' , 'Very High'])
```

```python
df['Education_level'] = df['Education'].map({1: 'Below College' , 2: 'College' , 3: 'Bachelor' , 4: 'Master' , 5: 'Doctor'})
```

```python
df['JobSatisfaction_level'] = df['JobSatisfaction'].map({1: 'Low' , 2: 'Medium' , 3: 'High' , 4: 'Very High'})
```

```python
df['WorkLifeBalance_level'] = df['WorkLifeBalance'].map({1: 'Bad' , 2: 'Good' , 3: 'Better' , 4: 'Best'})
```
